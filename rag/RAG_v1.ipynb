```python
from sentence_transformers import SentenceTransformer
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# model = SentenceTransformer('all-MiniLM-L6-v2')
model = SentenceTransformer("BAAI/bge-m3") # Best model right now!
```

```python
commands = [
    "command", 
    "takeoff", 
    "land", 
    "up", 
    "down", 
    "left", 
    "right",
    "forward", 
    "back", 
    "battery"
]
```

```python
# Exactly the same as the commands list from "Tello SDK 2.0 User Guide"
# plain_descriptions = [
#     "Enter SDK mode.", 
#     "Auto takeoff.", 
#     "Auto landing.", 
#     "Ascend to ‘x’ cm.",
#     "Descend to ‘x’ cm.", 
#     "Fly left for ‘x’ cm.", 
#     "Fly right for ‘x’ cm.",
#     "Fly forward for ‘x’ cm.", 
#     "Fly backward for ‘x’ cm.", 
#     "Obtain current battery percentage."
# ]

# Refined descriptions for each command
descriptions = [
    "Enter SDK mode to begin interacting with the drone.",  # Refined description for 'command'
    "Initiate automatic takeoff, making the drone ascend to a preset height.",  # Refined for 'takeoff'
    "Initiate automatic landing, bringing the drone safely down to the ground.",  # Refined for 'land'
    "Ascend the drone vertically to a specific height of ‘x’ cm.",  # Refined for 'up'
    "Descend the drone vertically to a specific height of ‘x’ cm.",  # Refined for 'down'
    "Move the drone left by ‘x’ cm from its current position.",  # Refined for 'left'
    "Move the drone right by ‘x’ cm from its current position.",  # Refined for 'right'
    "Move the drone forward by ‘x’ cm along its current direction.",  # Refined for 'forward'
    "Move the drone backward by ‘x’ cm along its current direction.",  # Refined for 'back'
    "Retrieve and display the current battery percentage of the drone."  # Refined for 'battery'
]
```

```python
description_embeddings = model.encode(descriptions)
```

```python
def find_best_match(query):
    # Convert query into embedding
    query_embedding = model.encode([query])
    
    # Calculate cosine similarity between query and descriptions
    similarities = cosine_similarity(query_embedding, description_embeddings)
    
    # Find the index of the highest similarity
    best_match_idx = np.argmax(similarities)
    
    # Return the most similar command and its details
    return commands[best_match_idx], descriptions[best_match_idx]
```

```python
for user_query, expected_command in questions_with_expected:
    best_command, best_description = find_best_match(user_query)
    is_correct = "✅ Correct" if best_command == expected_command else f"❌ Incorrect (Expected: {expected_command})"
    print(f"Question: {user_query}")
    print(f"Best Match Command: {best_command}")
    print(f"Description: {best_description}")
    print(f"Result: {is_correct}")
    print("-" * 60)
```
